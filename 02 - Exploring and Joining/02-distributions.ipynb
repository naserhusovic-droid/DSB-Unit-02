# Scenario

You're an analyst working for CoolBricks - a new toy company who wants to compete with LEGO by producing their own bricks and sets. To better understand what kind of sets to create and at what price points, you are tasked with analyzing LEGO's catalogue over time.

The data comes from Kaggle: [https://www.kaggle.com/datasets/alexracape/lego-sets-and-prices-over-time](https://www.kaggle.com/datasets/alexracape/lego-sets-and-prices-over-time)

We are going to use `pandas` to read and explore this data visually.

The question we want to answer is: **what does a "typical" LEGO product look like**?

# Descriptive statistics
---

```python
# import pandas

import pandas as pd

lego_df = pd.read_csv("https://raw.githubusercontent.com/naserhusovic-droid/DSB-Unit-02/refs/heads/main/data/lego.csv")
```

Let's read in the data.

```python
lego_df.info()
```

Let's look at some summary statistics.

```python
lego_df.head(10)
```

```python
# kako mozemo dobiti vise statistickih podataka
# informacije o nasem data setu

lego_df["USD_MSRP"].describe()
```

The maximum value being so much higher than the 75th percentile tells us this data is probably skewed. Let's see:

```python
lego_df["USD_MSRP"].hist(bins=50)
```

And the same thing as a box plot.

```python
lego_df.boxplot("USD_MSRP", vert=False)
```

Let's look at the number of pieces now.

```python
lego_df["Pieces"].describe()
```

```python
lego_df.boxplot("Pieces", vert=False)
```

**Discussion:** What do these plots tells us about the "typical" number of pieces in a LEGO set?
